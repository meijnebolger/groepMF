## Processing
**Description of processing of raw data into scientific evidence:**


```python
#Data processing and analysis:

def analyse(data):
    N = len(data)
    gemiddelde = np.mean(data)
    std_dev = np.std(data, ddof=1)
    meetonzekerheid = std_dev / np.sqrt(N)
    relatieve = meetonzekerheid / gemiddelde * 100
    return gemiddelde, std_dev, meetonzekerheid, relatieve


gem_5, sd_5, u_5, r_5 = analyse(t_5)
gem_10, sd_10, u_10, r_10 = analyse(t_10)

print("5 metingen:")
print(f"Gemiddelde: {gem_5:.3f} s, Stdev: {sd_5:.3f} s, Meetonzekerheid: {u_5:.3f} s, Relatieve onzekerheid: {r_5:.2f}%.")

print("\n10 metingen:")
print(f"Gemiddelde: {gem_10:.3f} s, Stdev: {sd_10:.3f} s, Meetonzekerheid: {u_10:.3f} s, Relatieve onzekerheid: {r_10:.2f}%.")



#Hoe reduceert de meetonzekerheid door 10 periodes/valtijd te bepalen?
#De meetonzekerheid gaat van 0.070 naar 0.030 s wanneer het aantal metingen verdubbeld. Door 10 periodes te meten, wordt de invloed van toevallige fouten zoals reactietijd gemiddeld over een langere tijd. Hierdoor neemt de relatieve fout af.
# Is de verkregen meetonzekerheid in de tijd voldoende om te bepalen met een foutenmarge van 0.5%? Nee

# Histogram van 10 metingen
plt.hist(t_10, bins=5, edgecolor='black')
plt.xlabel('Valtijd (s)')
plt.ylabel('Frequentie')
plt.title('Verdeling van 10 vrije valmetingen')
plt.show()

h = 1.0

# Bereken valversnelling g = 2h / t^2
g = 2 * h / (gem_5 ** 2)
print(g)

```
