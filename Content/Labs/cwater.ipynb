# Bepaling van soortelijke warmte van water


# Introductie

Water heeft een enorme buffercapaciteit: je moet veel energie toevoegen om de temperatuur van water een graad te verwarmen. In dit practicum gaan we de soortelijke warmte van water bepalen door een bekende hoeveelheid water te verwarmen met een bekende hoeveelheid energie, en de temperatuurstijging te meten.

# Theorie

Warmte kan worden overgedragen tussen stoffen of voorwerpen als gevolg van temperatuurverschil.
Warmte is een vorm van energie en geeft een maat voor de gemiddelde kinetische energie van moleculen.

De hoeveelheid energie die nodig is om 1 kilogram van een stof 1 Kelvin in temperatuur te laten stijgen
is de soortelijke warmte. Water staat bekend om zijn relatief hoge soortelijke warmte.

## Formules

1. **Toegevoerde energie:**

$$
Q = U \cdot I \cdot t
$$

2. **Soortelijke warmte:**

$$
c = \frac{Q}{m \cdot \Delta T}
$$

Waarbij:

- **Q** = toegevoerde warmte (J)  
- **m** = massa (g)  
- **c** = soortelijke warmte (J/g·K)  
- **$\Delta T$** = temperatuurverandering (K)

De toegevoerde energie wordt verkregen door gebruik te maken van een verwarmingselement, ofwel de toegevoerde energie kan ook worden genoteerd volgens:

$$
Q = P \cdot t
$$

met:

- **P** = vermogen van het verwarmingselement (W)  
- **t** = tijd (s)

Deze afwijking kan worden verklaard door warmteverlies aan de omgeving, onnauwkeurigheden in de temperatuurmeting en de aanname dat alle elektrische energie volledig wordt omgezet in warmte van het water.

# Resultaten


````python
#stappenplan

#1. start met het meten van de temperatuur van het water
#2. zet de voedingsbron op een geschikte spanning en stroom, voor het verwarwarmingselement is dat bekend en 12 V en 2 ampere
#3. leg het verwarmingselement in het water en start de timer
#4. verwarm tot een duidelijke temperatuurstijging en stop de timer (denk aan 4 tot 5 graden celcius)
#5. Ga met deze resultaten na wat de soortelijke warmte is van water. Gebruik hiervoor de volgende formules:

# Berekening soortelijke warmte van water
# Formules: Q = U * I * t, c = Q / (m * ΔT)

# Gegeven waarden (voorbeeld)
U = 12.0        # spanning in volt
I = 1.32         # stroom in ampère
m = 0.520        # massa water (kg)
T_begin = 18.9   # vul eigen gemeten waarde in (°C)
T_eind = 20.8   # vul gemeten eindwaarden in (°C)
t = 220     #vul gemeten tijd in

# Berekeningen
Q = U * I * t  # energie in joule
delta_T = T_eind - T_begin
c = Q / (m * delta_T)

# Resultaten tonen
print(f"Toegevoegde energie Q = {Q:.2f} J")
print(f"Temperatuurstijging ΔT = {delta_T:.2f} °C")
print(f"Soortelijke warmte c = {c:.2f} J/(kg·°C)")
````
