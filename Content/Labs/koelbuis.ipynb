# Theorie

Volgens [Newton's wet van afkoeling](https://nl.wikipedia.org/wiki/Afkoelingswet_van_Newton) is de snelheid waarmee een voorwerp afkoelt evenredig met het verschil in de temperatuur van het voorwerp ($T$) en de omgeving ($T_0$). We kunnen dit schrijven als:

$$
    \dot{Q} = -hA(T(t) - T_0),
$$

waarin 

- $\dot{Q}$ de warmtestroom in $\mathrm{W}$,
- $A$ het oppervlak waardoor koeling optreedt in $\mathrm{m}^2$,
- $h$ de warmteoverdrachtscoëfficiënt in $\mathrm{W/(m^2 K)}$.

Dit levert de differentiaalvergelijking

$$
    C\dot{T} = -hA(T(t) - T_0),
$$

met $C$ de warmtecapaciteit in $\mathrm{J/kg}$. Herschrijven met $\tau = \frac{C}{hA}$ levert:

$$
    -\tau\dot{T} = T(t) - T_0,
$$ (eq:dv1)

met als oplossing:

$$
    T(t) - T_0 = (T(0) - T_0)\text{e}^{-t/\tau}.
$$ (eq:dv1_opl)

We kunnen hieruit dus concluderen dat $\tau$ de karakteristieke tijdsduur is waarin de temperatuur van de buis een factor $\text{e}$ verlaagd ten opzichte van de omgevingstemperatuur.

We zijn hier voor het gemak uitgegaan van een $h$ die onafhankelijk is van de temperatuur. We weten echter dat warmtetransport door straling niet lineair gaat, maar als 

$$
    \dot{Q}_s = \epsilon \sigma A (T^4 - T_0^4).
$$ 

Voor kleine temperatuurverschillen ($\Delta T = T - T_0$) is dit te vereenvoudigen tot

$$
    \dot{Q}_s = \epsilon \sigma A ((T_0+\Delta T)^4 - T_0^4) \approx \epsilon \sigma 4A T_0^3 \Delta T.
$$

Zolang $\Delta T$ dus relatief klein is ten opzichte van $T_0$, kunnen we $h$ dus inderdaad als een constante beschouwen. 

```{exercise} Uitwerken
:label: ex_uitw

Laat zien dat bovenstaande geldt door eerst $(T_0+\Delta T)^4$ uit te schrijven, te substitueren en dan te bedenken dat $\Delta T$ klein is en $\Delta T^2$ dus nog kleiner.
```

```{solution} ex_uitw

$$
\dot{Q}_s = \epsilon \sigma A (T^4 - T_0^4)
$$

\(\Delta T = T - T_0\), 
dus \(T = T_0 + \Delta T\).
 \((T_0 + \Delta T)^4\):

$$
(T_0 + \Delta T)^4 = T_0^4 + 4 T_0^3 \Delta T + 6 T_0^2 (\Delta T)^2 + 4 T_0 (\Delta T)^3 + (\Delta T)^4
$$

Dan substitueren in \(\dot{Q}_s\):

$$
\dot{Q}_s = \epsilon \sigma A \left(4 T_0^3 \Delta T + 6 T_0^2 (\Delta T)^2 + 4 T_0 (\Delta T)^3 + (\Delta T)^4\right)
$$

Stap 3: kleine \(\Delta T\) benadering (\(\Delta T \ll T_0\)):

$$
\dot{Q}_s \approx \epsilon \sigma A \, 4 T_0^3 \Delta T
$$Hiermee is het warmtetransport lineair in \(\Delta T\) en kunnen we een constante \(h = 4 \epsilon \sigma T_0^3\) gebruiken.

```


```{exercise} 
:label: ex_fout
Hoe groot is de fout in het warmtetransport door straling die we maken voor het temperatuurbereik waarin we deze proef uitvoeren?
```

```{solution} ex_fout
De relatieve fout van de linearisatie:

$$
\text{Fout} = \frac{6 T_0^2 (\Delta T)^2 + 4 T_0 (\Delta T)^3 + (\Delta T)^4}{4 T_0^3 \Delta T} \approx \frac{6 T_0^2 (\Delta T)^2}{4 T_0^3 \Delta T} = \frac{3}{2} \frac{\Delta T}{T_0}
$$

Bijvoorbeeld, voor \(\Delta T = 10~\mathrm{K}\) en \(T_0 = 300~\mathrm{K}\):

$$
\text{Fout} \approx 1.5 \frac{10}{300} = 0.05 = 5\%
$$

De fout is dus klein voor kleine \(\Delta T\) en de linearisatie is geldig.

Maakt het uit in welke richting je de (cilindrische) as van de buis legt? En waarom zit er een dop bij de buis?

```{solution} ex_conf
De buis is cillindrisch en koelt af via convectie en straling. Convectie, warmteoverdracht naar lucht, gebeurd tussen de verwarmde buis en de omgeven lucht rond de buis. Als de buis verticaal is georiënteerd kan de verwarmde lucht makkelijk opstijgen langs de buis en wordt die vervangen door koelere lucht, hierdoor koelt de buis sneller af. Gezien het horizontaal plaatsen van de buis ervoor zorgt dat de warme lucht minder goed kan opstijgen, hierdoor wordt de afkoeling meer geremd. 

De dop die op de buis wordt geplaatst sluit een zijde van de buis af. Dit zorgt ervoor dat warmte binnenin de buis niet makkelijk kan onstnappen. De warmte wordt gedwongen om via het buitenoppervlak van de buis afgevoerd te worden. De afkoeling loopt langzamer en meer gelijkmatig. De dop zorgt dus voor een controleerbare afkoeling.

```

# Resultaten

`````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

times = np.array([0, 30, 60, 90, 120, 150, 180, 210, 240, 270, 300, 330, 360, 390, 420, 450, 480, 510, 540, 570, 600, 630, 660])  # tijd in seconden

temps = np.array([53.3, 51.0, 48.9, 47.4, 45.8, 44.2, 42.6, 41.6, 40.6, 39.5, 38.7, 37.7, 36.9, 36.2, 35.5, 34.7, 34.0, 33.6, 33.1, 32.7, 32.2, 31.8, 31.3]) + 273.15  # omzetting naar Kelvin

buitenoppervlak = 0.0187      # m^2
warmtecapaciteit = 880        # J/K


def cooling_model(t, A, tau, T_omg):
    return A * np.exp(-t / tau) + T_omg

# Beginwaarden 
A0 = temps[0] - temps[-1]     # startverschil
tau0 = 700                    # s
T_omg0 = temps[-1]            # benadering omgeving

popt, pcov = curve_fit(cooling_model, times, temps, p0=[A0, tau0, T_omg0], maxfev=5000)

A_fit, tau_fit, T_omg_fit = popt

print(f"Karakteristieke tijd τ = {tau_fit:.1f} s")
print(f"Omgevingstemperatuur = {T_omg_fit - 273.15:.1f} °C")

# Warmteoverdrachtscoëfficiënt
h = warmtecapaciteit / (tau_fit * buitenoppervlak)
print(f"Warmteoverdrachtscoëfficiënt h = {h:.1f} W/(m² K)")

t_fit = np.linspace(0, times.max(), 300)
T_fit = cooling_model(t_fit, *popt)

plt.figure()
plt.scatter(times, temps - 273.15, label="Meting")
plt.plot(t_fit, T_fit - 273.15, label="Exponentiële fit")
plt.xlabel("Tijd (s)")
plt.ylabel("Temperatuur (°C)")
plt.legend()
plt.show()
`````

# Discussie en conclusie

We verwachten dat het grootste deel van de warmtedrachtscoefficient gegeven is door convectie, doordat warmte door luchtstroming wordt afgekoeld. Straling speelt ook een rol, maar speelt een minder grote rol dan convectie. Van geleiding is er hier het minst sprake, aangezien er weinig contactpunten zijn.
