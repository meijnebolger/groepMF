# Waxinekaarsje

Als je een waxinekaarsje aan steekt en er een glas overheen zet, zal het kaarsje uit gaan. 
Als je dat waxinekaarsje laat drijven op water terwijl je er het glas er overheen zet, zal het water in het glas stijgen.
Op het internet zwerven er verschillende verklaringen voor dit fenomeen rond.
1. Het kaarsje verbruikt zuurstof, waardoor er een vacuüm ontstaat en het water omhoog wordt gezogen.
2. Het kaarsje verwarmt de lucht in het glas, waardoor de lucht uitzet. Wanneer de kaars dooft koelt de lucht af waardoor de druk afneemt en er water omhoog wordt gezogen.
3. Bij het verbranden van de kaars ontstaan er waterdamp en koolstofdioxide. De waterdamp condenseert aan de binnenkant van het glas, waardoor er minder gas in het glas is en de druk afneemt. Hierdoor wordt het water omhoog gezogen.

```{exercise}
Onderzoek welke verklaring het meest waarschijnlijk is en leg uit waarom de andere verklaringen minder waarschijnlijk zijn.
```

1. Verklaring 1:

Ja het waxinelichtje verbruikt zuurstof, maar niet genoeg om te zorgen voor een durkverschil groot genoeg om water op te zuigen. Een vacuüm ontstaat als het aantal gasmoleculen in een afgesloten ruimte daalt, waardoor de druk lager wordt dan buiten. Maar gezien de verbruikte zuurstofdeeltjes vervangen worden voor de verbrandingsproducten (CO₂ en waterdamp), blijft de dichtheid van de deeltjes vrijwel hetzelfde. Er onstaat geen groot genoeg vacuüm die het water zover omhoog kan zuigen.

2. Verklaring 2:

Om op de proef te stellen dat de verwarmde lucht in het glas ervoor zorgt dat het water wordt opgezogen zodra het afkoelt doen we de volgende proef:

Door de lucht al voor te verwarmen in de afgesloten glazen buis, kunnen we de invloed van de afgekoelde lucht testen. Het kaarsje drijft in het water maar nu wordt de glazen buis erop gezet met al voorverwarmde lucht door het met een aansteker te verwarmen.

Met de al verwarmde lucht in de glazen buis zou het water minder moeten stijgen, omdat de lucht minder sterk afkoelt. Het drukverschil is dan kleiner en het water wordt minder omhoog gezogen. 

De uitzetting van de lucht door het verwarmen van de vlam van het waxinelichtje is dus een verklaring voor het onstaan van een vacuüm als de lucht daarna weer afkoelt. 

3. Verklaring 3:

In de vorige proef gebruikte we de al voorverwarmde lucht in de glazen buis om te verklaren wat de invloed was van het vormen van een vacuüm door het afkoelen van warme lucht.

Met deze proef valt ook verklaring 3 te onderzoeken.Wawterdamp condenseert tegen het koude binnenkant van de glazen buis. De binnenkant van het glas is koeler dan de hete lucht vlakbij de vlam van het waxinelichtje. Het waterdamp dat onstaat bij het verbranden van kaarsvet verandert dan van gas naar vloeistof. Dit betekent dat er minder gasmoleculen overblijven en er een vacuüm ontstaat dat het water omhoog zuigt.

Wanneer de binnenkant van de glazenbuis al is voorverwarmd, door het opwarmen van de lucht in de buis, condenseert er minder waterdamp en wordt het drukverschil minder groot. Hierdoor stijgt het water minder.


CONCLUSIE

De meest waarschijnlijke verklaring voor dit fenomeen is dat de uitzetting van de lucht ervoor zorgt dat wanneer deze lucht afkoelt er een vacuüm ontstaat, ofwel verklaring 2. Gezien de condensatie van waterdamp in dit experiment minimaal is en daarom weinig bijdraagt in het vormen van een vacuüm en het zuurstofverbruik van een waxinelichtje nauwelijks effect heeft op de druk binnen de buis.
