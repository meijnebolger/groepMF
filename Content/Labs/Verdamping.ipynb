# Is het verdamping?

In dit experiment is er een verwarmingselement in een met water gevulde maatbeker gestopt. Elke minuut is de temperatuur van het water gemeten. Deze metingen zijn opgeslagen in [tempmetingen.csv](tempmetingen.csv).

Verder is gegeven dat: 
$m_{maatbeker} = 820.8 \mathrm{gr}$  
$m_{maatbeker + water, start} = 1292.9 \mathrm{gr}$  
$m_{maatbeker + water, eind} = 1274.9 \mathrm{gr}$

Het moge duidelijk zijn dat er water is 'verdwenen'. Ook is de eindtemperatuur van het water niet gelijk aan een proces waarbij verdamping en eventuele warmteverliezen niet meegenomen worden.

````python
import numpy as np
import matplotlib.pyplot as plt
metingen = np.loadtxt(r'C:\Users\meijn\Downloads\tempmetingen-46931600a9ec9432720d281e3972dd50.csv', delimiter=';', skiprows=1)

tijd = metingen[:, 0]
temp = metingen[:, 1]

m_maat = 820.8 #gram

#massa van water berekenen
m_begin = 1292.9 - m_maat #gram
m_eind = 1274.9 - m_maat #gram

#trendlijn maken
#we gaan ervan uit dat alle toegevoegde warmte uitsluitend water opwarmt, dit is een lineaire relatie, we halen de helling uit het beginstuk van de metingen, waar de verdamping nog minimaal is. delta t in 10 min = 13.5 graden

T0 = temp[0]
helling = 1.35  # °C per minuut
trend = [T0 + helling * t for t in tijd]

plt.plot(tijd, temp)
plt.plot(tijd, trend)
plt.title("Grafiek temperatuurmetingen en verwachte trendlijn")
plt.xlabel('Tijd min')
plt.ylabel('Temperatuur in C')
````

````python

c = 4180          # J/(kg*K), soortelijke warmte water
Lv = 2.26e6       # J/kg, verdampingswarmte water

# Massa's (kg)
m_water_start = 0.472
m_water_eind = 0.454
m_water_gem = (m_water_start + m_water_eind) / 2
m_verdampt = 0.018

# Temperaturen (graden)
T_start = 21.2
T_eind = 76.6
delta_T = T_eind - T_start

# Energieën
Q_opwarming = m_water_gem * c * delta_T
Q_verdamping = m_verdampt * Lv
Q_totaal = Q_opwarming + Q_verdamping

# Vermogen
tijd = 60 * 60  # seconden
P = Q_totaal / tijd

Q_opwarming, Q_verdamping, Q_totaal, P
````

Hierboven is de energie die nodig is vooor het opwarmen van het water en de energie die nodig is voor het verdampen van water ebrekend, door deze twee bij elkaar op te tellen krijg je een schatting van de totale toegevoegde energie. Met deze energie kom je op een vermogen van 40 W gedurende 60 min, dit is realistisch voor een klein verwarmingselement. De toegevoegde energie is dus consistent met de opwarming en verdamping van het water.

Verbetering experiment:
In het experiment zouden warmteverliezen kunnen worden verminderd, bijvoorbeled door een isolerende mantel rond de maatbeker te plaatsen, of een dekel te gebruiken met maar een klein gat voor het verwarmingselement. Hierdoor wordt het een meer geisoleerd systeem en kun je beter berekenen hoeveel energie naar verdamping gaat.
